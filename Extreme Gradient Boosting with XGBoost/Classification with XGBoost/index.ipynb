```python
class_data = pd.read_csv('../resources/breast-cancer.csv')

class_data['diagnosis'] = np.where(class_data['diagnosis'] == 'M' ,1,0)

X = class_data.drop('diagnosis', axis=1).to_numpy()
y_labels = class_data['diagnosis'].to_numpy()


X_train, X_test, y_train, y_test = train_test_split(X, y_labels, test_size=0.2, stratify=y_labels, random_state=123)

xg_cl = xgb.XGBClassifier(objective='binary:logistic',n_estimators=10,seed=123)

xg_cl.fit(X_train,y_train)

y_pred = xg_cl.predict(X_test)

accuracy_score(y_test,y_pred)
```

# Descision Trees as a base learners
- Base learner - Individual learning algorithm in an ensemble algorithm
- Composed of series of binary questions
- Predictions happen at the 'leaves' of the tree


- A decision tree is built iteratively, making one binary split at a time until a stopping condition is reached (such as maximum depth or pure nodes). At each step, the algorithm chooses the best feature and split point that separates the target classes more clearly. The goal is to group similar target values together so that each branch becomes increasingly dominated by a single class. This process continues until the data in each final node (leaf) mostly or completely belongs to one category, allowing the tree to make predictions based on the majority class in that leaf.


# CART : Classification and Regression Trees

- Each leaf always contains a real-valued score
- Can later be converted into categories
- A real-valued score is a numerical prediction produced by the tree instead of a class label.

# What is Boosting

### Boosting Overview
- Not a specific machine learning algorithm
- Concepts that can be applied to a set of machine learning models 'Meta-Algorithm'
- Ensemble algoritm used to convert many weak learners

### Weak learners and strong learners
- **Weak Learner** : ML algorithm that is slightly better than chance 
    - Example : Descision Tree whose predictions are slightly better than 50%
- XGboost converts a collection of weak learners into a strong learner

- **Strong learner** : Any algorithm that can be tuned to achieve good performance

# How boosting is accomplished
- Iterativelly learning a set of weak models on subets of the data
- Weighing each weak prediction according to each weak learner's performance
- Combine the weighted predictions to obtain a single weighted prediction
- that is much better than the individual predictions themselves

# Model evaluation through cross-validation
- **Cross-Validation** : Robust method for estimating the performnce of a model on unseen data
- Generates many non-overlapping train/test splits on training data
- Reports the average test performance across all data splits

# Cross Val in XGBoost 

```python
class_dmatrix = xgb.DMatrix(data=X, label=y_labels)
params = {"objective":'binary:logistic','max_depth':4}

cv_results = xgb.cv(dtrain=class_dmatrix,params=params,nfold=10,num_boost_round=20,metrics='error',as_pandas=True)

accuracy = 1 - cv_results["test-error-mean"].iloc[-1]

print(f"Accuracy {accuracy} ")
```

# What to use XGBoost 
- You have a large number of training samples 
    - Greater than 1000 training samples and less 100 features 
    - The number of features < number of training samples

- You have a mixture of categorical and numeric features
