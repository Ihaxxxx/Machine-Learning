```python
import xgboost as xgb 
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv('../resources/housing.csv')
```

# Untuned Model

```python
X , y = df.iloc[:,:-1] , df.iloc[:,-1]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=123
)

housing_dmatrix = xgb.DMatrix(data=X,label=y)

untuned_params = {'objective':'reg:squarederror'}

untuned_cv_results_rmse = xgb.cv(dtrain=housing_dmatrix,nfold=4,params=untuned_params,as_pandas=True,seed=123,metrics='rmse')

# print("Untuned rmse : %f" % ((untuned_cv_results_rmse["test-rmse-mean"]).tail(1)))
untuned_cv_results_rmse["test-rmse-mean"].tail(1)
```

# Tuned example

```python
X , y = df.iloc[:,:-1] , df.iloc[:,-1]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=123
)

housing_dmatrix = xgb.DMatrix(data=X,label=y)

tuned_params = {'objective':'reg:squarederror','colsample_bytree':0.3,'learning_rate':0.1,'max_depth':5}

tuned_cv_results_rmse = xgb.cv(dtrain=housing_dmatrix,nfold=4,params=tuned_params,num_boost_round=200,as_pandas=True,seed=123,metrics='rmse')

tuned_cv_results_rmse["test-rmse-mean"].tail(1)
```
