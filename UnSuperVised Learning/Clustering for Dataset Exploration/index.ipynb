# 1. Unsupervised Learning
- Learning patterns without labels.
- Examples:
- Clustering: Group similar items together.
- Dimension reduction: Compress data while keeping patterns.

# 2. Supervised vs Unsupervised
- Feature	Supervised Learning	Unsupervised Learning
- Labels	Uses labels (guided)	No labels (unguided)
- Goal	Predict output	Discover patterns
- Example	Tumor classification	Customer clustering

3. Iris Dataset
- Contains 3 species of iris plants.
- Features: Sepal Length, Sepal Width, Petal Length, Petal Width.
- Data shape: (samples, features) → (rows, columns).
- 4-dimensional: Each sample is a 4D point.

4. k-Means Clustering
- Finds k clusters in data.
- Uses centroids (mean of cluster points).
- New points assigned to closest centroid.

```python
# 1. Import libraries
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import numpy as np

# 2. Load iris dataset
iris = load_iris()
X = iris.data  # shape (150, 4)
print("First 5 samples:\n", X[:5])

# 3. Create k-means model (3 clusters)
kmeans = KMeans(n_clusters=3, random_state=42)

# 4. Fit model to the data
kmeans.fit(X)

# 5. Get cluster labels
labels = kmeans.labels_
print("Cluster labels:\n", labels)

# 6. Predict clusters for new samples
new_samples = np.array([[5.0, 3.5, 1.5, 0.2], [6.0, 3.0, 5.0, 2.0]])
predicted_labels = kmeans.predict(new_samples)
print("Predicted cluster labels for new samples:", predicted_labels)

# 7. Scatter plot (Sepal length vs Petal length)
plt.scatter(X[:,0], X[:,2], c=labels, cmap='viridis')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.title('Iris Clusters (Sepal vs Petal length)')
plt.show()
```
