# One hot encoding

```python
pd.get_dummies(df,columns=['Country'],prefix='C')
```

# Dummy Encoding

```python
pd.get_dummies(df,columns=['Country'],drop_first=True,prefix='C')
# out = df.filter(regex='^C')
# print
```

# One - hot vs Dummies
- **One hot encoding** : Explainable features
- **Dummy Encoding** : Necessary information without duplication

# Limiting Columns

```python
counts = df['Country'].value_counts()
print(counts)

mask = df['Country'].isin(counts[counts < 20].index)
df.loc[mask,'Country'] = 'Other'
print(df['Country'].value_counts())
```

```python
import pandas as pd

data = {
    "Restaurant_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Number_of_Violations": [5, 2, 8, 1, 0, 4, 7, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
df['Binary_Violation'] = 0

df.loc[df['Number_of_Violations'] > 0 , "Binary_Violation"] = 1

print(df)
```

```python
df['Binned_Group'] = pd.cut(
    df['Number_of_Violations'],
    bins=[-np.inf,0,2,np.inf],
    labels=[1,2,3]
)

df
```
