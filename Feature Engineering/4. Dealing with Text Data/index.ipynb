```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
speech = pd.read_csv('../resources/us.csv')
train_speech = pd.read_csv('../resources/us.csv')
test_speech = pd.read_csv('../resources/us.csv')
print(speech.info())
```

```python
speech.head()
```

```python
speech['text'] = speech['text'].str.replace('[^a-zA-Z]', ' ', regex=True)
speech['text'] = speech['text'].str.lower()
speech['text'] = speech['text'].str.replace(r'\s+', ' ', regex=True)  # remove extra spaces
speech['text'] = speech['text'].str.strip()

print(speech['text'].head())
```

```python
speech['char_cnt'] = speech['text'].str.len()

speech['char_cnt'].head()
```

```python
print(speech['text'].head())
speech['text_clean'] = speech['text']
speech['text_clean'].head()
```

```python
speech['word_cnt'] = speech['text'].str.split().str.len()
speech['word_cnt'].head()
```

```python
speech['avg_word_len'] = speech['char_cnt'] / speech['word_cnt']
```

```python
speech['avg_word_len'].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(min_df=0.1,max_df=0.9)
```

```python
cv.fit(speech['text_clean'])
```

```python
cv_transformed = cv.transform(speech['text_clean'])
print(cv_transformed)

cv_transformed.toarray()

feature_name = cv.get_feature_names_out()
print(feature_name)
```

```python
cv_df = pd.DataFrame(
    cv_transformed.toarray(),
    columns=cv.get_feature_names_out()
).add_prefix("Counts_")

print(cv_df.head())
```

```python
speech = pd.concat([speech,cv_df],axis=1,sort=False)

print(speech.head())
print(speech.shape)
```

# Term frequency-inverse document frequency

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tv = TfidfVectorizer(stop_words='english')
print(tv)
```

# Max features and stopwords

```python
tv.fit(train_speech['text'])
train_tv_transformed = tv.transform(train_speech['text'])
```

```python
train_tv_df = pd.DataFrame(
    train_tv_transformed.toarray(),
    columns = tv.get_feature_names_out()
).add_prefix("TFIDF_")

train_speech_df = pd.concat([train_speech,train_tv_df],axis=1,sort=False)
examine_row = train_tv_df.iloc[0]


print(examine_row.sort_values(ascending=False))
```

```python
# transform the test set (use test_speech, not train_speech)
test_tv_transformed = tv.transform(test_speech['text'])
test_tv_df = pd.DataFrame(
    test_tv_transformed.toarray(),
    columns = tv.get_feature_names_out()
).add_prefix("TFIDF_")

test_speech_df = pd.concat([test_speech,test_tv_df],axis=1,sort=False)

# examine only the TFIDF (numeric) columns to avoid comparing strings with floats
examine_row = test_tv_df.iloc[0]

print(examine_row.sort_values(ascending=False))
```
