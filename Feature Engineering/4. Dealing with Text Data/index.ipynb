```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
speech = pd.read_csv('../resources/us.csv')
print(speech.info())
```

```python
speech.head()
```

```python
speech['text'] = speech['text'].str.replace('[^a-zA-Z]',' ',regex=True)
speech['text'].head()
speech['text'] = speech['text'].str.lower()
print(speech['text'])
```

```python
speech['char_cnt'] = speech['text'].str.len()

speech['char_cnt'].head()
```

```python
speech['text_clean'] = speech['text'].str.replace('[^a-zA-Z]',' ',regex=True).str.split(" ")
```

```python
speech['word_cnt'] = speech['text'].str.split().str.len()
speech['word_cnt'].head()
```

```python
speech['avg_word_len'] = speech['char_cnt'] / speech['word_cnt']
```

```python
speech['avg_word_len'].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(min_df=0.1,max_df=0.9)
```

```python
speech['text_clean']
```

```python
cv_transformed = cv.fit_transform(speech['text_clean'])
print(cv_transformed)
```
