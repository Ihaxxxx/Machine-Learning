# Remove Missing data

```python
df = pd.DataFrame({
    'A': [10, np.nan, 30, 40, np.nan, 60, 70],
    'B': [5, 15, np.nan, 35, 45, 55, np.nan],
    'C': [np.nan, 22, 33, np.nan, 55, 66, 77]
})
```

```python
print(df.dropna())
```

# Drop Specific Rows

```python
print(df.drop([1,2,3]))
```

# Drop Columns

```python
print(df.drop("A",axis=1))
```

# Drop Rows with NAN

```python
df.isna().sum()
```

```python
df.dropna(subset=["B"])
```

# Working with data types

### Why are types important

```python
print(volunteer.info())
```

```python
df = pd.DataFrame({
    'A': [10, 5, 30],
    'B': ["string", "string", "string"],
    'C': [2, 10, 33]
})
```

```python
print(df.info())
print(df)
```

```python
df['B'] =  df["B"].astype("string")
df['A'] =  df["A"].astype("float")
df['C'] =  df["C"].astype("float")

print(df.dtypes)
```
