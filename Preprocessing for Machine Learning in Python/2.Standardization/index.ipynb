# What is standardization

- **Standardization** : transform continuos data to appear normally distributed
- This is necessary step because many models make underlying assumptions that the training data is normally distributed and if it isn't we could risk biasing your models .
- log normalization and feature scalling in this course
- Applied to continuos numerical data

# When to standardize : linear distances
- Model in linear space
- Examples :
    - K-Nearest Neighbours (kNN)
    - Linear Regression
    - K-Means Clustering
- Standardization should also be used when datasets features have a high variance , which is also related to distance metrics
- This could bias a model that assumes the data is normally distributed . If a feature in our data set has a variance that's and order of magnitude or more greater than the other features
- This could impact the model's abilit to learn from other features in the dataset

# What is feature scalling
- Scalling is a mehtod of standardization that's most useful when we're working with a data set that contains continuos features that are on different scales
- Model which uses linear space for example linear regression and K-nearest neighbors .
- Features are transformed so that the features in our data set have a mean of zero and a variance of 1 . 
- Transforms to normal distribution

```python
import pandas as pd

df = pd.DataFrame({
    'Age': [18, 21, 25, 30, 28, 24, 22],            # Scale ~ 20s
    'Salary': [25000, 32000, 28000, 35000, 30000, 27000, 31000],  # Scale ~ 30,000
    'Distance_m': [1, 3, 2, 5, 4, 2, 3]             # Scale ~ 1–5
})

print(df)
print(df.var())
```

# How to scale data

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df),columns=df.columns)

print(df_scaled)
print(df_scaled.var())
```
