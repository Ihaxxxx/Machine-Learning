```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Categorical plots
- involve a categorial variable
- Comparisions between groups

```python
tips = sns.load_dataset("tips")
```

```python
df = pd.read_csv('../resources/mpg.csv')
df.info()
```

```python
sns.countplot(
    x='origin',
    data=df
)
plt.show()
```

```python
sns.catplot(
    x='origin',
    data=df,
    kind='count',
    hue='origin'
)
plt.show()
```

# Change order

```python
cat_or = ["usa",'japan','europe']
```

```python
sns.catplot(
    x='origin',
    data=df,
    kind='count',
    hue='origin',
    order=cat_or
)
```

# Bar plots

```python
sns.catplot(
    x='origin',
    y='horsepower',
    data=df,
    kind='bar',
    hue='origin',
)
```

```python
sns.catplot(
    x='origin',
    y='horsepower',
    data=df,
    kind='bar',
    hue='origin',
    errorbar=None
)
```

```python
sns.catplot(
    y='origin',
    x='horsepower',
    data=df,
    kind='bar',
    hue='origin',
    errorbar=None
)
```

# Box plots
- Shows the distribution of quantative data
- See median spread skewness and outliers
- Facilitates comparisions between groups

```python
sns.catplot(
    x='time',
    y='total_bill',
    data=tips,
    kind='box'
)
```

```python
sns.catplot(
    x='time',
    y='total_bill',
    data=tips,
    kind='box',
    order=['Dinner','Lunch']
)
```

```python
sns.catplot(
    x='time',
    y='total_bill',
    data=tips,
    kind='box',
    showfliers=False
)
```

```python
sns.catplot(
    x='time',
    y='total_bill',
    data=tips,
    kind='box',
    showfliers=False,
    whis=2
)
```

```python
sns.catplot(
    x='time',
    y='total_bill',
    data=tips,
    kind='box',
    showfliers=False,
    whis=[5,95]
)
```

# Point plots
- points show mean of quanatative variable
- Verticle lines show 95% confidence intervals



### Point plots vs Line plots
- **Similarities**
- Mean of quantative variable
- 95% confidence intervals for the mean

- **Differences**
- Line plot has quantative variable (usually time) on x-axis
- Point plot has categorical variables on x-axis


```python
tips.head()
```

```python
sns.catplot(
    x='size',
    y='total_bill',
    data=tips,
    hue='sex',
    kind='point'
)
```

```python
sns.catplot(
    x='size',
    y='total_bill',
    data=tips,
    hue='sex',
    kind='point',
    linestyle=None
)
```

```python
from numpy import median

sns.catplot(
    x='smoker',
    y='total_bill',
    data=tips,
    kind='point',
    estimator=median
)
```

```python
sns.catplot(
    x='smoker',
    y='total_bill',
    data=tips,
    kind='point',
    capsize=0.2
)
```

```python
sns.catplot(
    x='smoker',
    y='total_bill',
    data=tips,
    kind='point',
    errorbar=None

)
```
