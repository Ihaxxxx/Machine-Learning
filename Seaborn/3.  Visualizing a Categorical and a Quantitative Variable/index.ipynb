```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Categorical plots
- involve a categorial variable
- Comparisions between groups

```python
df = pd.read_csv('../resources/mpg.csv')
df.info()
```

```python
sns.countplot(
    x='origin',
    data=df
)
plt.show()
```

```python
sns.catplot(
    x='origin',
    data=df,
    kind='count',
    hue='origin'
)
plt.show()
```

# Change order

```python
cat_or = ["usa",'japan','europe']
```

```python
sns.catplot(
    x='origin',
    data=df,
    kind='count',
    hue='origin',
    order=cat_or
)
```

# Bar plots

```python
sns.catplot(
    x='origin',
    y='horsepower',
    data=df,
    kind='bar',
    hue='origin',
)
```

```python
sns.catplot(
    x='origin',
    y='horsepower',
    data=df,
    kind='bar',
    hue='origin',
    errorbar=None
)
```

```python
sns.catplot(
    y='origin',
    x='horsepower',
    data=df,
    kind='bar',
    hue='origin',
    errorbar=None
)
```
