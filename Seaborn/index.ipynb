```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Scatter Plot

```python

height = np.array([150, 155, 160, 162, 168, 170, 172, 175, 180, 185])
weight = np.array([50, 54, 58, 60, 65, 68, 70, 74, 80, 85])
gender = np.array([
    "Male", "Female", "Male", "Female", "Male",
    "Female", "Male", "Male", "Female", "Female"
])


sns.scatterplot(x=height,y=weight)
plt.show()
```

# Count plot

```python
sns.countplot(x=gender)
plt.show()
```
