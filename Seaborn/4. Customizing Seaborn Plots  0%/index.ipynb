```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../resources/countries-of-the-world.csv')
```

```python
tips = sns.load_dataset('tips')
```

```python
sns.set_palette
```

# Facet grid

```python
g = sns.catplot(
    x='Region',
    y="Birthrate",
    data=df,
    kind='box'
)

g.figure.suptitle("New Title",y=1.03)
```

# AxesSubplot

```python
g = sns.boxplot(
    x='Region',
    y='Birthrate',
    data=df
)

g.set_title("sdakdas",y=1.04)
plt.xticks(rotation=90)
```
