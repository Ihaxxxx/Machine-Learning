# Introduction to relational plots and subplots

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
tips = sns.load_dataset('tips')
tips.head()
```

```python
hue_colors = {
    "Yes" : "black",
    "No" : "red"
}
sns.scatterplot(
    x='total_bill',
    y='tip',
    hue='smoker',
    data=tips,
    hue_order=['Yes','No'],
    palette=hue_colors
)

plt.show()
```

# Using relplot

```python
sns.relplot(
    x='total_bill',
    y='tip',
    data=tips,
    kind='scatter',
    col='smoker',
    row='time'
)

plt.show()
```

```python
sns.relplot(
    x='total_bill',
    y='tip',
    data=tips,
    kind='scatter',
    col='day',
    col_wrap=2,
    col_order = ['Thur','Fri','Sat','Sun']
)

plt.show()
```

# Customizing scatter plots

```python
tips.head()
```

```python
sns.relplot(
    x='total_bill',
    y='tip',
    data=tips,
    kind='scatter',
    size='size',
    hue='size'
)
plt.show()
```

```python
sns.relplot(
    x='total_bill',
    y='tip',
    data=tips,
    kind='scatter',
    hue='smoker',
    style='smoker'
)
plt.show()
```

```python
sns.relplot(
    x='total_bill',
    y='tip',
    data=tips,
    kind='scatter',
    alpha=0.4
)
plt.show()
```

# Line plots

- Each point represents the same "thing" typically tracked over time

```python
df = pd.read_csv("../resources/air_quality_hourly_dataset.csv")
```

```python
df.head()
```

```python

df['datetime'] = pd.to_datetime(df['datetime'])
filtered = df[(df['datetime'] >= '2024-01-01') & 
              (df['datetime'] <= '2024-01-02')]
```

```python


sns.relplot(
    x='datetime',
    y='NO2_mean',
    data=filtered,
    kind='line'
)

plt.show()
```

```python


sns.relplot(
    x='datetime',
    y='NO2_mean',
    data=filtered,
    kind='line',
    hue='location',
    style='location',
    markers=True,
    dashes=False
)

plt.show()
```

```python
sns.relplot(
    x='datetime',
    y='NO2_mean',
    data=filtered,
    kind='line',
    errorbar='sd'
)
```
